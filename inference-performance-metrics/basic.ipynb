# Inference Performance Metrics - Basic

Core metrics for evaluating inference systems: latency, throughput, memory, and quality.

```python
from __future__ import annotations

from dataclasses import dataclass

@dataclass(frozen=True)
class InferenceMetrics:
    latency_ms: float
    throughput_rps: float
    memory_mb: float
    quality_score: float

reference = InferenceMetrics(latency_ms=38.4, throughput_rps=120.0, memory_mb=842.0, quality_score=0.91)
candidate = InferenceMetrics(latency_ms=31.2, throughput_rps=138.0, memory_mb=910.0, quality_score=0.89)
reference, candidate
```

```python
def summarize_tradeoff(baseline: InferenceMetrics, variant: InferenceMetrics) -> dict[str, float]:
    return {
        "latency_delta_ms": baseline.latency_ms - variant.latency_ms,
        "throughput_delta_rps": variant.throughput_rps - baseline.throughput_rps,
        "memory_delta_mb": variant.memory_mb - baseline.memory_mb,
        "quality_delta": variant.quality_score - baseline.quality_score,
    }

summary = summarize_tradeoff(reference, candidate)
summary
```

## Summary

Use latency, throughput, memory, and quality together when selecting an inference configuration.
