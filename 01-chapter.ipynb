```python
import os
os.environ["KERAS_BACKEND"] = "tensorflow"

import tensorflow as tf
import keras_hub
import numpy as np

# Create custom dummy dataset to train the base model
X_base = np.array(["great movie", "awful film", "i loved it", "terrible script"])
y_base = np.array([1, 0, 1, 0])

# --- REBUILD BERT TINY FROM SCRATCH ---
# load_weights=False skips downloading the pre-trained parameters!
base_classifier = keras_hub.models.BertTextClassifier.from_preset(
    "bert_tiny_en_uncased", 
    num_classes=2,
    load_weights=False  # <--- CRITICAL SWITCH
)

base_classifier.compile(
    optimizer=tf.keras.optimizers.AdamW(learning_rate=1e-4),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=["accuracy"]
)

print("=== Phase 1: Training the custom Base Model from absolute zero ===")
base_classifier.fit(X_base, y_base, epochs=3, batch_size=2)

# Save the freshly randomized/trained base model structure
base_classifier.save("my_custom_scratch_bert")
```
