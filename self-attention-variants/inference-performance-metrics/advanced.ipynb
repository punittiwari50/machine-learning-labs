# Inference Performance Metrics - Advanced

Serving-scale metrics for long-context and generative inference systems.

```python
from __future__ import annotations

from dataclasses import dataclass

@dataclass(frozen=True)
class ServingProfile:
    p95_latency_ms: float
    ttf_token_ms: float
    tokens_per_second: float
    kv_cache_mb: float
    cost_per_1k_tokens: float

dense = ServingProfile(142.0, 54.0, 28.0, 1840.0, 0.83)
gqa = ServingProfile(96.0, 37.0, 41.0, 1120.0, 0.61)
dense, gqa
```

```python
def compare_profiles(baseline: ServingProfile, optimized: ServingProfile) -> dict[str, float]:
    return {
        "p95_latency_ms_saved": baseline.p95_latency_ms - optimized.p95_latency_ms,
        "ttf_token_ms_saved": baseline.ttf_token_ms - optimized.ttf_token_ms,
        "throughput_gain": optimized.tokens_per_second - baseline.tokens_per_second,
        "kv_cache_mb_saved": baseline.kv_cache_mb - optimized.kv_cache_mb,
        "cost_per_1k_tokens_saved": baseline.cost_per_1k_tokens - optimized.cost_per_1k_tokens,
    }

comparison = compare_profiles(dense, gqa)
comparison
```

```python
benchmark_axes = {
    "tail_latency": (dense.p95_latency_ms, gqa.p95_latency_ms),
    "time_to_first_token": (dense.ttf_token_ms, gqa.ttf_token_ms),
    "tokens_per_second": (dense.tokens_per_second, gqa.tokens_per_second),
    "kv_cache_mb": (dense.kv_cache_mb, gqa.kv_cache_mb),
}
benchmark_axes
```

## Summary

Advanced serving decisions should balance tail latency, decode speed, cache footprint, and unit cost.
