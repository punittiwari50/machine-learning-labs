# 04 - Embeddings Detailed Guide

This notebook covers practical embedding concepts with examples and enterprise-focused patterns.

## 1) Types of Embeddings in ML (Technical Categories + Functional Use Cases)

Embeddings convert sparse or unstructured inputs into dense vectors so ML models can compute semantic similarity, ranking, clustering, and retrieval efficiently.

### A) Token-Level Static Embeddings
- **Definition**: One fixed vector per token regardless of context.
- **Examples**: Word2Vec, GloVe, FastText.
- **Functional use case**: Lightweight NLP pipelines where speed and low memory matter.
- **Real-time example**: Keyword expansion in e-commerce search (`tv` ~ `television`).

### B) Contextual Token Embeddings
- **Definition**: Token vectors change based on surrounding words.
- **Examples**: BERT, RoBERTa, DeBERTa.
- **Functional use case**: High-accuracy NER, QA, sentiment in ambiguous text.
- **Real-time example**: Distinguish `bank` in `river bank` vs `bank account`.

### C) Sentence / Document Embeddings
- **Definition**: One vector for an entire sentence or chunk.
- **Examples**: Sentence-BERT, E5, MiniLM embeddings.
- **Functional use case**: Semantic search, RAG retrieval, duplicate detection.
- **Real-time example**: Match `refund not received` to `payment reversal pending`.

### D) Domain-Specific Embeddings
- **Definition**: Embeddings trained/fine-tuned on domain corpora.
- **Examples**: BioClinicalBERT embeddings, FinBERT-style domain vectors.
- **Functional use case**: Better precision in legal/medical/finance language.
- **Real-time example**: Insurance claim routing using policy-specific terminology.

### E) Multilingual / Cross-Lingual Embeddings
- **Definition**: Shared semantic space for multiple languages.
- **Examples**: LASER, multilingual MiniLM/E5 variants.
- **Functional use case**: Cross-language retrieval and support operations.
- **Real-time example**: Spanish query retrieves English knowledge base article.

### F) Multimodal Embeddings
- **Definition**: Different modalities (text, image, audio) embedded into one space.
- **Examples**: CLIP-like text-image embeddings.
- **Functional use case**: Search and recommendation across content types.
- **Real-time example**: Search product images using text (`red running shoes`).

### G) Graph / Entity Embeddings
- **Definition**: Encode nodes/relations from knowledge graphs.
- **Examples**: TransE, Node2Vec, GraphSAGE embeddings.
- **Functional use case**: Link prediction, fraud ring detection, recommendation.
- **Real-time example**: Detect suspicious transaction entities via graph proximity.

### H) Positional Embeddings
- **Definition**: Encode token position so models know word order.
- **Examples**: Sinusoidal position encoding (Transformer), learned positional vectors.
- **Functional use case**: Sequence understanding in Transformers.
- **Real-time example**: Differentiate `dog bites man` vs `man bites dog`.

### I) Relative / Rotary Positional Terms
- **Definition**: Encode relative distance or rotational position in attention.
- **Examples**: Relative position bias, RoPE (Rotary Position Embeddings), ALiBi.
- **Functional use case**: Better long-context handling.
- **Real-time example**: Long contract Q&A where clause distance matters.

### J) Segment / Token-Type Embeddings
- **Definition**: Mark which segment/token group a token belongs to.
- **Examples**: BERT token-type embeddings for sentence A/B.
- **Functional use case**: Sentence-pair classification and QA.
- **Real-time example**: Distinguish question tokens from passage tokens in enterprise search.

### K) Patch / Feature Embeddings (Vision and Time-Series)
- **Definition**: Convert image patches or feature windows into vectors.
- **Examples**: ViT patch embeddings, time-series patch tokenization.
- **Functional use case**: Bring non-text inputs into transformer pipelines.
- **Real-time example**: Defect detection from camera frames in manufacturing.

```python
from math import sin, cos
import numpy as np

def cos_sim(a, b):
    a = np.array(a, dtype=float)
    b = np.array(b, dtype=float)
    na = np.linalg.norm(a)
    nb = np.linalg.norm(b)
    return 0.0 if na == 0 or nb == 0 else float(np.dot(a, b) / (na * nb))

print("1) Semantic search (sentence embedding style)")
q = [0.85, 0.15, 0.25]
d1 = [0.84, 0.16, 0.26]
d2 = [0.2, 0.8, 0.2]
print("best=doc_1" if cos_sim(q, d1) > cos_sim(q, d2) else "best=doc_2")

print("\n2) Recommendation (static token embedding)")
static_emb = {"shoe": [0.9, 0.2, 0.1], "sneaker": [0.88, 0.21, 0.1]}
print("shoe~sneaker=", round(cos_sim(static_emb["shoe"], static_emb["sneaker"]), 4))

print("\n3) Intent routing (contextual style output)")
contextual = {"bank@river": [0.2, 0.8, 0.1], "bank@finance": [0.9, 0.1, 0.2]}
print("bank sense similarity=", round(cos_sim(contextual["bank@river"], contextual["bank@finance"]), 4))

print("\n4) Duplicate detection (sentence vector pooling)")
tok = {"refund": [0.8, 0.2, 0.2], "not": [0.1, 0.1, 0.1], "received": [0.75, 0.2, 0.3]}
sent = np.mean(np.array([tok["refund"], tok["not"], tok["received"]]), axis=0).tolist()
sent2 = [0.56, 0.17, 0.2]
print("dup_score=", round(cos_sim(sent, sent2), 4))

print("\n5) Domain-specific retrieval")
finance_emb = {"invoice": [0.95, 0.1, 0.1], "claim": [0.9, 0.15, 0.12]}
print("invoice~claim=", round(cos_sim(finance_emb["invoice"], finance_emb["claim"]), 4))

print("\n6) Cross-lingual retrieval")
multi = {"hola": [0.82, 0.2, 0.1], "hello": [0.81, 0.21, 0.1]}
print("hola~hello=", round(cos_sim(multi["hola"], multi["hello"]), 4))

print("\n7) Multimodal retrieval")
text_vec = [0.7, 0.2, 0.6]
image_vec = [0.68, 0.18, 0.59]
print("text~image=", round(cos_sim(text_vec, image_vec), 4))

print("\n8) Graph/entity risk-linking")
node_emb = {"user_1": [0.9, 0.1, 0.2], "merchant_9": [0.86, 0.12, 0.22]}
print("user_1~merchant_9=", round(cos_sim(node_emb["user_1"], node_emb["merchant_9"]), 4))

print("\n9) Positional embedding (token + position)")
token = [0.6, 0.3, 0.2]
pos_2 = [0.03, 0.01, 0.02]
print("composed=", [round(x, 3) for x in (np.array(token) + np.array(pos_2)).tolist()])

print("\n10) Relative/Rotary position idea")
x = np.array([1.0, 0.0])
theta = 0.5
rot = np.array([[cos(theta), -sin(theta)], [sin(theta), cos(theta)]]) @ x
print("rotated=", [round(v, 3) for v in rot.tolist()])

print("\n11) Token-type/segment embedding")
seg_A = [0.01, 0.0, 0.0]
seg_B = [0.0, 0.01, 0.0]
print("token+A=", (np.array([0.4, 0.5, 0.1]) + np.array(seg_A)).round(3).tolist())
print("token+B=", (np.array([0.4, 0.5, 0.1]) + np.array(seg_B)).round(3).tolist())

print("\n12) Patch embedding (vision/time-series)")
patch = np.array([12, 20, 18, 5], dtype=float)
W = np.array([[0.1, 0.2, 0.0], [0.0, 0.1, 0.2], [0.2, 0.0, 0.1], [0.1, 0.1, 0.1]])
patch_vec = (patch @ W).tolist()
print("patch_vec=", [round(v, 3) for v in patch_vec])
```

```python
# Example: token embedding + positional embedding composition
tokens = ["man", "bites", "dog"]

token_embed = {
    "man":   [0.8, 0.1, 0.2],
    "bites": [0.4, 0.9, 0.1],
    "dog":   [0.75, 0.15, 0.25],
}

# Learned positional embeddings for positions 0, 1, 2
pos_embed = {
    0: [0.01, 0.02, 0.03],
    1: [0.02, 0.01, 0.04],
    2: [0.03, 0.01, 0.02],
}

composed = []
for i, tok in enumerate(tokens):
    # Transformer input is typically token_vec + position_vec
    vec = [a + b for a, b in zip(token_embed[tok], pos_embed[i])]
    composed.append((tok, i, vec))

for tok, pos, vec in composed:
    print(f"token={tok:5} pos={pos} composed_vec={vec}")

print("\nUse case: Order-sensitive meaning in sequence models (chat, contracts, logs).")
```

```python
# Use-case-first short snippets
import numpy as np

def cosine_sim(a, b):
    a = np.array(a, dtype=float)
    b = np.array(b, dtype=float)
    na = np.linalg.norm(a)
    nb = np.linalg.norm(b)
    return 0.0 if na == 0 or nb == 0 else float(np.dot(a, b) / (na * nb))

# 1) Semantic search
q = np.array([0.9, 0.1, 0.2])
docs = {"refund_policy": [0.88, 0.12, 0.22], "shipping": [0.15, 0.85, 0.2]}
best_doc = max(docs, key=lambda k: cosine_sim(q, docs[k]))
print("semantic_search ->", best_doc)

# 2) Intent routing
intents = {"billing": [0.95, 0.1, 0.2], "tech": [0.1, 0.95, 0.2]}
msg = [0.92, 0.12, 0.21]
best_intent = max(intents, key=lambda k: cosine_sim(msg, intents[k]))
print("intent_routing ->", best_intent)

# 3) Multilingual retrieval
multi = {"hola": [0.81, 0.2, 0.1], "hello": [0.8, 0.21, 0.1]}
print("cross_lingual(hola,hello)=", round(cosine_sim(multi["hola"], multi["hello"]), 4))

# 4) Multimodal retrieval
text_vec = [0.7, 0.2, 0.6]
img_vec = [0.69, 0.19, 0.61]
print("cross_modal(text,image)=", round(cosine_sim(text_vec, img_vec), 4))

# 5) Graph proximity
node = {"acct_1": [0.9, 0.2, 0.1], "acct_2": [0.88, 0.22, 0.12]}
print("graph_similarity=", round(cosine_sim(node["acct_1"], node["acct_2"]), 4))
```

```python
from math import sqrt

# Tiny toy embeddings (3 dimensions for visibility)
E = {
    "king":   [0.9, 0.8, 0.2],
    "queen":  [0.88, 0.79, 0.21],
    "man":    [0.7, 0.2, 0.1],
    "woman":  [0.69, 0.21, 0.12],
    "apple":  [0.1, 0.92, 0.85],
    "banana": [0.09, 0.9, 0.82],
}

def dot(a, b):
    return sum(x * y for x, y in zip(a, b))

def norm(v):
    return sqrt(sum(x * x for x in v))

def cosine(a, b):
    na, nb = norm(a), norm(b)
    if na == 0 or nb == 0:
        return 0.0
    return dot(a, b) / (na * nb)

pairs = [("king", "queen"), ("apple", "banana"), ("king", "apple")]
for a, b in pairs:
    print(f"cos({a},{b}) = {cosine(E[a], E[b]):.4f}")
```

```python
# Example: simple nearest-neighbor lookup
def top_k_similar(word, k=3):
    base = E[word]
    scores = []
    for w, vec in E.items():
        if w == word:
            continue
        scores.append((w, cosine(base, vec)))
    scores.sort(key=lambda x: x[1], reverse=True)
    return scores[:k]

print("Top similar to 'king':", top_k_similar("king"))
print("Top similar to 'apple':", top_k_similar("apple"))
```

### Code Explanation: Similarity and Nearest Neighbors

What this code does technically:
- Creates a tiny embedding space (`E`) where each word is a dense vector.
- Computes cosine similarity to measure angular closeness in vector space.
- Returns top-k nearest neighbors as semantic alternatives.

Why this matters functionally:
- This is the core primitive used in semantic search and recommendation.
- If two vectors are close, the system treats the corresponding items as semantically related.

Simple use case:
- Input term: `king`
- Output: nearest semantic terms, useful for query expansion or related-item suggestions.

Real-time production analogue:
- User types `invoice correction`; system retrieves related intents like `billing update` and `payment amendment` via nearest vectors.

```python
# Mini real-time style matching for helpdesk intents
intent_vectors = {
    "billing_issue": [0.95, 0.10, 0.20],
    "password_reset": [0.15, 0.96, 0.18],
    "shipping_delay": [0.20, 0.25, 0.94],
}

incoming_query_vec = [0.93, 0.11, 0.21]

best_intent = max(intent_vectors, key=lambda k: cosine(incoming_query_vec, intent_vectors[k]))
score = cosine(incoming_query_vec, intent_vectors[best_intent])
print(f"Routed intent: {best_intent}, similarity={score:.4f}")
```

## 4) Special Edge Cases

Key edge cases to handle in production:

- **Out-of-vocabulary (OOV)** tokens for static embeddings.
- **Short queries** (e.g., "it", "error") that are ambiguous.
- **Very long documents** where chunking strategy changes retrieval quality.
- **Multilingual/code-mix input** leading to poor neighborhood quality.
- **Domain drift** where vocabulary changes over time (new product names).
- **Near-duplicate vectors** creating unstable ranking ties.
- **PII leakage risk** if sensitive text is embedded and stored without policy.

```python
# Edge case helper: safe cosine + threshold gate
def safe_route(query_vec, candidates, threshold=0.80):
    if not query_vec or norm(query_vec) == 0:
        return "fallback_human", 0.0

    best = max(candidates, key=lambda k: cosine(query_vec, candidates[k]))
    score = cosine(query_vec, candidates[best])
    if score < threshold:
        return "fallback_human", score
    return best, score

print(safe_route([0.0, 0.0, 0.0], intent_vectors))
print(safe_route([0.1, 0.1, 0.1], intent_vectors, threshold=0.95))
```

## 5) Do and Don't

**Do**

- Use cosine similarity for directional semantic comparison.
- Add threshold + fallback path (human/escalation/rules).
- Re-embed when model version changes, and keep version metadata.
- Evaluate with domain-specific test sets (not only generic benchmarks).
- Monitor drift and retrieval quality over time.

**Don't**

- Do not assume high similarity means factual correctness.
- Do not mix embeddings from different models in same vector index.
- Do not ignore privacy/compliance constraints for stored vectors.
- Do not skip chunking strategy design for long documents.
- Do not deploy without offline + online evaluation gates.

## 7) Enterprise Code Sample with Explanation

Below is a simple enterprise-style semantic retrieval pipeline skeleton:

- Encodes text into vectors (placeholder encoder).
- Stores vectors with metadata including access policy and model version.
- Applies policy filter before ranking.
- Uses threshold gate for safe fallback.

### 7.1 Enterprise Short Code (Functional Pattern)

Minimal enterprise flow: embed -> ACL filter -> rank -> threshold fallback.

```python
# Enterprise short pattern: ACL + threshold gate
import numpy as np

def cs(a, b):
    a = np.array(a, dtype=float)
    b = np.array(b, dtype=float)
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))

def embed(text):
    text = text.lower()
    return np.array([len(text) % 10, text.count("r"), text.count("e")], dtype=float)

docs = [
    {"id": "fin-1", "acl": "finance", "vec": embed("refund processing guide")},
    {"id": "fin-2", "acl": "finance", "vec": embed("invoice correction policy")},
    {"id": "it-1", "acl": "it", "vec": embed("password reset steps")},
]

query = embed("refund not received")
allowed = [d for d in docs if d["acl"] == "finance"]
ranked = sorted([(d["id"], cs(query, d["vec"])) for d in allowed], key=lambda x: x[1], reverse=True)
top_id, top_score = ranked[0]
print(("fallback_human", round(top_score, 3)) if top_score < 0.8 else (top_id, round(top_score, 3)))
```

```python
from dataclasses import dataclass
from typing import List, Dict

@dataclass
class DocRecord:
    doc_id: str
    text: str
    vector: List[float]
    acl: str
    model_version: str

class EnterpriseVectorStore:
    def __init__(self):
        self.docs: List[DocRecord] = []

    def add(self, rec: DocRecord):
        self.docs.append(rec)

    def search(self, query_vec: List[float], user_acl: str, top_k: int = 3, threshold: float = 0.80):
        # Policy first: only rank docs user is allowed to see.
        scoped = [d for d in self.docs if d.acl == user_acl]
        scored = [(d, cosine(query_vec, d.vector)) for d in scoped]
        scored.sort(key=lambda x: x[1], reverse=True)

        top = [(d.doc_id, s, d.model_version) for d, s in scored[:top_k] if s >= threshold]
        if not top:
            return {"action": "fallback", "reason": "low_similarity_or_no_access"}

        return {"action": "results", "items": top}

# Placeholder deterministic encoder (replace with real model in production).
def pseudo_encoder(text: str) -> List[float]:
    t = text.lower()
    return [
        (sum(ord(ch) for ch in t[:10]) % 100) / 100.0,
        (len(t) % 100) / 100.0,
        (sum(ch in "aeiou" for ch in t) % 100) / 100.0,
    ]

store = EnterpriseVectorStore()
store.add(DocRecord("doc-001", "How to request invoice correction", pseudo_encoder("invoice correction"), "finance", "embed-v1"))
store.add(DocRecord("doc-002", "Password reset policy", pseudo_encoder("password reset"), "it", "embed-v1"))
store.add(DocRecord("doc-003", "Refund processing SLA", pseudo_encoder("refund processing"), "finance", "embed-v1"))

query = "refund not received"
qvec = pseudo_encoder(query)
print(store.search(qvec, user_acl="finance", top_k=2, threshold=0.65))
print(store.search(qvec, user_acl="it", top_k=2, threshold=0.65))
```

### Enterprise Sample Explanation

Why this pattern works in real systems:

- **Security by design**: ACL filtering before ranking avoids unauthorized retrieval.
- **Operational safety**: threshold gate prevents low-confidence results.
- **Reproducibility**: `model_version` lets you track re-embedding migrations.
- **Extensible**: you can swap `pseudo_encoder` with any managed embedding model API.

Production upgrades to consider:

- ANN index (FAISS/Milvus/Pinecone/OpenSearch vector) for scale.
- Hybrid retrieval (BM25 + vector rerank).
- PII masking before embedding.
- A/B testing for model versions and thresholds.
